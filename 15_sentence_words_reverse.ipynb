# **Question 6. Write a function called reverse_words that takes a sentence as input and returns the sentence with the order of words reversed. Each word in the sentence should remain unchanged, but the order of the words should be reversed.**

```python
def reverse_words(sentence):
    words_list = sentence.split(" ")
    reverse_list = words_list[::-1]
    sentence = " ".join(reverse_list)
    return sentence
```

```python
sentence = "I am, a good boy"
reverse_words(sentence)
```

```python
sentence = "How are you?"
reverse_words(sentence)
```

# ALTERNATIVELY

```python
def reverse_words(sentence):
    s = ""
    sentence += " " # adding space at the end of a sentence

    # storing each word in a list
    words_list = []
    for ch in sentence:
        if ch == " ":
            words_list.append(s)
            s = ""
        else:
            s += ch
            
    # reversing words_list
    i = 0
    j = len(words_list)-1
    while i<j :
        temp = words_list[i]
        words_list[i] = words_list[j]
        words_list[j] = temp
        i += 1
        j -= 1

    return " ".join(words_list)
```

```python
sentence = "I am, a good boy"
reverse_words(sentence)
```

```python
sentence = "How are you?"
reverse_words(sentence)
```
