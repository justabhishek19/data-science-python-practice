#  Write python program to find the area of a triangle when 3 side of triangle given.

**<u>Herons's Formula:</u>**

Area = $\sqrt{s(s-a)(s-b)(s-c)}$

**a,b,c** are sides of a Triangle and $s = \frac{a+b+c}{2}$

```python
import numpy as np

def triangle_area(a,b,c):
    s = (a+b+c)/2

    area = np.sqrt(s*(s-a)*(s-b)*(s-c))

    return np.round(area,1)
    
```

```python
a = float(input('length of side a:'))
b = float(input('length of side b:'))
c = float(input('length of side c:'))
print(f'Area of Triangle: {triangle_area(a,b,c)}')
```
