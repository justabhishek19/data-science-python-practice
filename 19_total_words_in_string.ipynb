# Write a Python function that takes a string as input and returns the number of words in the string. Assume that words in the string are separated by spaces.

```python
def count_words(string):
    prev = string[0]
    count = 0
    for i in range(1,len(string)):
        current = string[i]
        if current == " " and prev != " ":
            count += 1
        prev = current

    if current == " ":
        count -= 1
        
    return count+1
```

```python
text = "This is a sample sentence."
count_words(text)
```

```python
text = "I  love       you  "
count_words(text)
```

# ALTERNATIVE

```python
def count_words(string):
    words = string.split()
    print(words)
    
    return len(words)
```

```python
text = "This is a sample sentence."
count_words(text)
```

```python
text = "I  love       you  "
count_words(text)
```
